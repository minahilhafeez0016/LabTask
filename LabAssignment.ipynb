# Question No. 1: Student Result Management System

```python
def calculate_grades(marks):
    total = sum(marks)
    percentage = (total / 300) * 100
    
    if percentage >= 80:
        grade = 'A'
    elif percentage >= 70:
        grade = 'B'
    elif percentage >= 60:
        grade = 'C'
    elif percentage >= 50:
        grade = 'D'
    else:
        grade = 'F'
        
    return total, percentage, grade

students_list = []

# Taking input for 5 students
print("--- Enter details for 5 students ---")
for i in range(5):
    print(f"\nStudent {i+1}:")
    name = input("Enter name: ")
    roll_no = input("Enter roll number: ")
    
    print("Enter marks out of 100 for:")
    python_marks = float(input("Python: "))
    ai_marks = float(input("Artificial Intelligence: "))
    math_marks = float(input("Mathematics: "))
    
    marks_list = [python_marks, ai_marks, math_marks]
    total, percentage, grade = calculate_grades(marks_list)
    
    student_dict = {
        "name": name,
        "roll_no": roll_no,
        "marks": marks_list,
        "total": total,
        "percentage": percentage,
        "grade": grade
    }
    students_list.append(student_dict)

# Displaying results
print("\n" + "="*40)
print("FINAL RESULTS OF ALL STUDENTS")
print("="*40)

highest_percentage = -1
top_student = ""

for s in students_list:
    print(f"Name: {s['name']} | Roll No: {s['roll_no']}")
    print(f"Total Marks: {s['total']}/300 | Percentage: {s['percentage']:.2f}% | Grade: {s['grade']}")
    print("-" * 40)
    
    if s['percentage'] > highest_percentage:
        highest_percentage = s['percentage']
        top_student = s['name']

print(f"\nTop Student: {top_student} with highest percentage of {highest_percentage:.2f}%")

# Saving to file
with open("results.txt", "w") as file:
    file.write("--- Student Results ---\n")
    for s in students_list:
        file.write(f"Name: {s['name']}, Roll No: {s['roll_no']}, Total: {s['total']}, Percentage: {s['percentage']:.2f}%, Grade: {s['grade']}\n")
    file.write(f"\nHighest Percentage Student: {top_student} ({highest_percentage:.2f}%)\n")

print("\nResults successfully saved to 'results.txt'!")
```

# Q#2: Sentence Analysis Program

```python
# Q#2: Sentence Analysis Program

sentence = input("Enter a sentence: ")

words = sentence.split()
total_words = len(words)
total_chars = len(sentence)

longest_word = max(words, key=len) if words else ""
shortest_word = min(words, key=len) if words else ""

vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in sentence if char in vowels)
unique_words = set(words)
sorted_words = sorted(words)

print("\n--- Sentence Analysis ---")
print(f"Total Words: {total_words}")
print(f"Total Characters: {total_chars}")
print(f"Longest Word: {longest_word}")
print(f"Shortest Word: {shortest_word}")
print(f"Number of Vowels: {vowel_count}")
print(f"Unique Words: {unique_words}")
print(f"Words in Alphabetical Order: {sorted_words}")

with open("word_analysis.txt", "w") as file:
    file.write(f"Sentence: {sentence}\n")
    file.write(f"Total Words: {total_words}\n")
    file.write(f"Total Characters: {total_chars}\n")
    file.write(f"Longest Word: {longest_word}\n")
    file.write(f"Shortest Word: {shortest_word}\n")
    file.write(f"Number of Vowels: {vowel_count}\n")
    file.write(f"Unique Words: {str(unique_words)}\n")
    file.write(f"Alphabetical Order: {str(sorted_words)}\n")

print("\nAnalysis successfully saved to 'word_analysis.txt'!")
```

# Q#3: Simple Shopping Program

```python
def calculate_bill(products):
    total_bill = 0
    detailed_bills = []
    
    for p in products:
        item_total = p['price'] * p['quantity']
        total_bill += item_total
        detailed_bills.append({'name': p['name'], 'item_total': item_total})
        
    discount = 0
    if total_bill > 10000:
        discount = total_bill * 0.10
        total_bill -= discount
        
    return total_bill, discount, detailed_bills

products_list = []
print("--- Enter 5 Products ---")
for i in range(5):
    print(f"\nProduct {i+1}:")
    name = input("Enter product name: ")
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))
    
    products_list.append({"name": name, "price": price, "quantity": quantity})

final_bill, discount_amount, itemized = calculate_bill(products_list)

# Find most expensive product by price
most_expensive = max(products_list, key=lambda x: x['price'])

print("\n" + "="*30)
print("BILL SUMMARY")
print("="*30)
for item in itemized:
    print(f"{item['name']}: Rs. {item['item_total']}")

if discount_amount > 0:
    print(f"\nDiscount Applied (10%): -Rs. {discount_amount}")
    
print(f"Final Total Bill: Rs. {final_bill}")
print(f"Most Expensive Product: {most_expensive['name']} (Rs. {most_expensive['price']})")

# Save bill to file
with open("bill.txt", "w") as file:
    file.write("--- Shopping Bill Summary ---\n")
    for item in itemized:
        file.write(f"{item['name']} - Total: Rs. {item['item_total']}\n")
    if discount_amount > 0:
        file.write(f"Discount: -Rs. {discount_amount}\n")
    file.write(f"Final Bill: Rs. {final_bill}\n")
    file.write(f"Most Expensive Product: {most_expensive['name']} (Rs. {most_expensive['price']})\n")

print("\nBill successfully saved to 'bill.txt'!")
```

# Q#4: Calculator Program with functions and loop

```python
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b): 
    return "Error! Division by zero." if b == 0 else a / b
def modulus(a, b): 
    return "Error! Modulus by zero." if b == 0 else a % b

while True:
    print("\n--- Simple Calculator ---")
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    
    print("Select operation:")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    print("5. Modulus (%)")
    
    choice = input("Enter choice (+, -, *, /, %): ")
    
    if choice == '+':
        result = add(num1, num2)
    elif choice == '-':
        result = subtract(num1, num2)
    elif choice == '*':
        result = multiply(num1, num2)
    elif choice == '/':
        result = divide(num1, num2)
    elif choice == '%':
        result = modulus(num1, num2)
    else:
        result = "Invalid Operation!"
        
    print(f"Result: {result}")
    
    cont = input("\nDo you want to perform another calculation? (yes/no): ").strip().lower()
    if cont != 'yes':
        print("Exiting calculator. Goodbye!")
        break
```

# Q#5: Small Restaurant Ordering Program

```python
menu = {
    1: {"item": "Burger", "price": 500},
    2: {"item": "Pizza", "price": 800},
    3: {"item": "Fries", "price": 250},
    4: {"item": "Drink", "price": 150}
}

order_list = []

print("--- Restaurant Menu ---")
for key, value in menu.items():
    print(f"{key}. {value['item']} - Rs. {value['price']}")

while True:
    choice = int(input("\nEnter item number you want to order (1-4): "))
    if choice in menu:
        qty = int(input(f"Enter quantity for {menu[choice]['item']}: "))
        item_total = menu[choice]['price'] * qty
        
        order_list.append({
            "item": menu[choice]['item'],
            "price": menu[choice]['price'],
            "quantity": qty,
            "total": item_total
        })
        print(f"Added {qty} {menu[choice]['item']}(s) to your order.")
    else:
        print("Invalid item choice!")
        
    more = input("Do you want to order more items? (yes/no): ").strip().lower()
    if more != 'yes':
        break

# Calculate final bill
grand_total = sum(item['total'] for item in order_list)

print("\n" + "="*30)
print("YOUR FINAL RESTAURANT BILL")
print("="*30)
for o in order_list:
    print(f"{o['item']} x {o['quantity']} = Rs. {o['total']}")
print("-" * 30)
print(f"Grand Total: Rs. {grand_total}")

# Save to orders.txt
with open("orders.txt", "w") as file:
    file.write("--- Restaurant Order Bill ---\n")
    for o in order_list:
        file.write(f"{o['item']} | Price: {o['price']} | Qty: {o['quantity']} | Total: {o['total']}\n")
    file.write(f"Grand Total: Rs. {grand_total}\n")

print("\nOrder bill successfully saved to 'orders.txt'!")
```
